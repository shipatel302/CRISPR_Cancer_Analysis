```python
# variables

class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def printPerson(self):
        return f"Person(name='{self.name}', age={self.age}, gender='{self.gender}')"


person_list = (Person("Priyesh", 30, "male"), Person("Shivani", 25, "female"))

for row in person_list:
    print(row.printPerson())


```

```python
def convert_ascii(val):
    return ord(val) - ord('0')

def convert_int(val):
    result = 0
    for i in val:
        result = result * 10 + convert_ascii(i)
    return result


print(type(convert_int("12")))  # Output: <class 'int'>

```
